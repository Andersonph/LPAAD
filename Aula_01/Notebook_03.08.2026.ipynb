# Problema 1

```python
# segunda = R$ 18.500,00
# terca = R$ 21.300,00
# quarta = R$ 17.800,00
# quinta = R$ 24.600,00
# sexta = R$ 27.400,00

fat_diario = [18500, 21300, 17800, 24600, 27400]

dias_analisados = len(fat_diario)
total_fat = sum(fat_diario)
media = total_fat / dias_analisados
maior_fat = max(fat_diario)
menor_fat = min(fat_diario)

print(f"Dias analisados: {dias_analisados}, Total de faturamento: {total_fat}, Faturamento médio: {media}, Maior faturamento registrado: {maior_fat}, Menor faturamento registrado: {menor_fat}.")

if media >= 22000:
    print("Bom desempenho")

if media >= 19000 and media < 22000:
    print("Desempenho regular")

if media < 19000:
    print("Desempenho crítico")
```
