```python
faturamentos = [18500, 22300, 19700, 25100, 1600, 21400]

meta = 20000

for faturamento in faturamentos:
    if faturamento >= meta:
        print(faturamento, "- Meta atingida")
    else:
        print(faturamento, "- Meta não atingida")
```

```python
notas = [8.5, 4.0, 7.5, 5.0, 9.0]

nota_maior = 0
cont = 0

for nota in notas:
    if nota >= 6:
        nota_maior += nota
        cont += 1

media = nota_maior / cont

print(f"{cont} notas foram maior que 6 e a média destas notas foi {media:.2f}.")
```
